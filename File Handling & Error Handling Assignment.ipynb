### Q6 : Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
def count_lines(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            line_count = sum(1 for line in file)
        return line_count
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None

filename = "`numbers.txt"
total = count_lines(filename)

if total is not None:
    print(f"Total line in file: {total}")
```

### Q14 :Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
try:
    num = int(input("Ek number enter karein: "))
    result = 10 / num
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Zero (0) se divide nahi ho sakta!")

except ValueError:
    print("Error: Sirf valid number hi enter karein!")

finally:
    print("Execution complete")
```

### Q17 : Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
try:
    num1 = int(input(" number enter karein: "))
    num2 = int(input(" number enter karein: "))
    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Kisi bhi number ko zero (0) se divide nahi kiya ja sakta!")

except ValueError:
    print("Error: Baraye meharbani sirf valid numbers (integers) enter karein!")
```

## Q18:`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    num1 = int(input(" number enter karein: "))
    num2 = int(input(" number enter karein: "))
    result = num1 / num2

except ZeroDivisionError:
    print("Error: Zero (0) se divide nahi kiya ja sakta!")

except ValueError:
    print("Error: Sirf valid numbers enter karein!")

else:
    print(f"Success! Answer hai: {result}")
```

### Q19 :`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Access granted! Aap eligible hain."

try:
    user_age = int(input("Apni age enter karein: "))
    result = check_age(user_age)
    print(result)
except ValueError as e:
    print(f"Error: {e}")    
```

### Q20 : File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file = None

try:
    file = open('sample.txt', 'r', encoding='utf-8')
    content = file.read()
    print("File Content:\n", content)

except FileNotFoundError:
    print("Error: 'sample.txt' file exist nahi karti.")

finally:
    if file and not file.closed:
        file.close()
        print("File successfully close kar di gayi hai.")
```
